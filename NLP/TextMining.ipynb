```python
import re
```

```python
text = "Contact          me on        info@gmail.com or call 66868687 #1234"
```

```python
#re.sub() -> remove,replace
#re.
```

Removing Number

```python
clean_text = re.sub(r'\d','',text)
clean_text
```

```python
remove_has = re.sub(r'#','',text)
remove_has
```

```python
remove_has_number = re.sub(r'#\d+','',text)
```

```python
remove_has_number
```

```python
new_text = "Contact me on info@gmail.com  abc@gmail.com  fkd@gmail.com or call 66868687 #312 890879"
```

```python
clean_new = re.sub(r'#\d+\s+\d+','',new_text)
clean_new
```

```python
space_remove = re.sub(r'\s+',' ',text)
```

```python
space_remove
```

Find

```python
find_call_num1 = re.findall(r'\w\s+\d+',space_remove)
find_call_num
```

```python
find_call_num = re.findall(r'call\s+\d+',space_remove)
find_call_num
```

```python
find_mail = re.findall(r'info@gmail.com',space_remove)
find_mail
```

```python
find_email = re.findall(r'\w+\@\w+\.\w+',space_remove)
find_email
```

```python
find_email = re.findall(r'\w+\@\w+\.\w+',new_text)
find_email
```

```python
for i in find_email:
    print(i)
```

```python
for e_idx,e_val in enumerate(find_email):
    print(f"{e_idx}:{e_val}")
```
